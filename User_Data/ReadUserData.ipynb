```python
import pickle

with open("dummy_users_v1.pkl", "rb") as f:
    dummy_users_loaded = pickle.load(f)
```

```python
for uid, ratings in dummy_users_loaded.items():
    liked = [rid for rid, rate in ratings.items() if rate == 1]
    disliked = [rid for rid, rate in ratings.items() if rate == -1]
    print(f"User {uid}:")
    print(f"  Liked recipes ({len(liked)}): {sorted(liked)}")
    print(f"  Disliked recipes ({len(disliked)}): {sorted(disliked)}\n")
```

```python
liked_data = []
disliked_data = []

for uid, ratings in dummy_users_loaded.items():
    for recipe_id, rating in ratings.items():
        if rating == 1:
            liked_data.append((uid, recipe_id))
        elif rating == -1:
            disliked_data.append((uid, recipe_id))
```

```python
def generate_sql_insert(table_name, data):
    values = ", ".join(f"({uid}, {rid})" for uid, rid in data)
    return f"INSERT INTO {table_name} (user_id, recipe_id) VALUES {values};"

liked_sql = generate_sql_insert("liked_recipes", liked_data)
disliked_sql = generate_sql_insert("disliked_recipes", disliked_data)

print(liked_sql)
print("----------------------------------------------")
print(disliked_sql)
```

```python
import random
from faker import Faker

fake = Faker()
Faker.seed(42)
random.seed(42)

sql_lines = ["INSERT INTO users (user_id, e_mail, user_name, user_bday, tel_no) VALUES"]

values = []
for user_id in range(1, 1001):
    email = fake.email().replace("'", "''")
    username = fake.user_name().replace("'", "''")
    bday = fake.date_of_birth(minimum_age=18, maximum_age=65).isoformat()
    
    # Generate a 10-digit integer phone number (leading digits won't be zero)
    phone = random.randint(1000000000, 9999999999)

    values.append(f"({user_id}, '{email}', '{username}', '{bday}', {phone})")

# Join and format
full_sql = "\n".join(sql_lines) + "\n" + ",\n".join(values) + ";"

# Save to file
with open("insert_dummy_users.sql", "w") as f:
    f.write(full_sql)

print("SQL insert statements with integer phone numbers saved to insert_dummy_users.sql")
```

```python
import random

sql_lines = ["INSERT INTO user_prefs (user_id, pref_id) VALUES"]

values = []

for user_id in range(1, 1001):
    num_prefs = random.randint(1, 3)
    prefs = random.sample(range(6), num_prefs)  # choose unique prefs
    for pref_id in prefs:
        values.append(f"({user_id}, {pref_id})")

# Join everything
full_sql = "\n".join(sql_lines) + "\n" + ",\n".join(values) + ";"

# Save to file
with open("insert_user_prefs.sql", "w") as f:
    f.write(full_sql)

print("User preferences SQL script written to insert_user_prefs.sql")
```
