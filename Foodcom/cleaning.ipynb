### Functinos

```python
def preprocess_text(text):
    if pd.isna(text):
        return ""
    text = text.lower().strip()  # Convert to lowercase and strip whitespace
    text = ''.join(e for e in text if e.isalnum() or e.isspace())  # Remove special characters
    return text
```

```python
def generate_recipe_hash(row):
    return hash((row['name'], tuple(sorted(row['ingredients'].split(','))), row['steps']))
```

```python
def process_chunk(chunk):
    """Process a single chunk for deduplication based on recipe hash."""
    chunk_copy = chunk.copy()  # Avoid modifying the original slice
    
    # Normalize text fields (add more preprocessing if needed)
    chunk_copy['name'] = chunk_copy['name'].apply(preprocess_text)
    chunk_copy['ingredients'] = chunk_copy['ingredients'].apply(preprocess_text)
    chunk_copy['steps'] = chunk_copy['steps'].apply(preprocess_text)
    
    # Generate hashes for exact duplicates
    chunk_copy['recipe_hash'] = chunk_copy.apply(generate_recipe_hash, axis=1)

    # Capture duplicates first, before removing them
    duplicates_with_reference = chunk_copy[chunk_copy['recipe_hash'].duplicated()]

    # Remove exact duplicates based on the hash and keep only the first
    chunk_copy = chunk_copy.drop_duplicates(subset=['recipe_hash'], keep='first')    

    return chunk_copy, duplicates_with_reference
```

```python
# Chunked processing
output_file = 'deduplicated_recipes.parquet'
chunk_size = 10000  # Adjust based on memory capacity
num_chunks = (len(mergedRecipes_df) + chunk_size - 1) // chunk_size  # Calculate total chunks
deduplicated_chunks = []
all_duplicates = []
```

```python
for i in range(num_chunks):
    print(f"Processing chunk {i + 1} of {num_chunks}...")
    chunk = mergedRecipes_df.iloc[i * chunk_size:(i + 1) * chunk_size]
    processed_chunk, duplicates = process_chunk(chunk)
    deduplicated_chunks.append(processed_chunk)
    all_duplicates.append(duplicates)
```

```python
# Combine all deduplicated chunks
deduplicated_df = pd.concat(deduplicated_chunks, ignore_index=True)

# Combine all duplicate rows for analysis
duplicates_df = pd.concat(all_duplicates, ignore_index=True)

# Save deduplicated data and duplicates
deduplicated_df.to_parquet('deduplicated_recipes.parquet', index=False)
duplicates_df.to_parquet('duplicates_recipes.parquet', index=False)

print("Deduplication complete! Results saved to:")
print(" - deduplicated_recipes.parquet (deduplicated data)")
print(" - duplicates_recipes.parquet (duplicates for review)")
```

```python
def save_to_csv_in_chunks(df, chunk_size, output_prefix):
    
    num_chunks = (len(df) + chunk_size - 1) // chunk_size  # Calculate total chunks
    for i in range(num_chunks):
        start_idx = i * chunk_size
        end_idx = min((i + 1) * chunk_size, len(df))
        chunk = df.iloc[start_idx:end_idx]
        output_filename = f"{output_prefix}_chunk_{i + 1}.csv"
        chunk.to_csv(output_filename, index=False)
        print(f"Saved {output_filename} with rows {start_idx} to {end_idx - 1}")
```
