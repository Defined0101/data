### Fix total time

```python
import re

def convert_to_minutes(duration):
    match = re.match(r'PT(?:(\d+)H)?(\d+)M', duration)
    if match:
        hours = int(match.group(1)) if match.group(1) else 0
        minutes = int(match.group(2))
        # Convert hours to minutes and add the minutes
        total_minutes = hours * 60 + minutes
        return total_minutes
    else:
        return None
```

```python
projectedIngr['TotalTime'] = projectedIngr['TotalTime'].apply(convert_to_minutes)
```

### Fix Quantities

```python
import re
from fractions import Fraction

def fix_quantities(quantities):
    cleaned_quantities = []
    for q in quantities:
        if q is None:
            return None
    
        if isinstance(q, str):

            q = q.replace('⁄', '/') #Replace Unicode character '⁄' (U+2044) with the regular slash '/'
            #print(q)

            # Handle ranges with mixed fractions and/or integers (e.g., '1 - 1 1/2', '1/3 - 1/2', '1/2 - 1')
            if re.match(r'(\d+/\d+|\d+\s+\d+/\d+|\d+)\s*-\s*(\d+/\d+|\d+\s+\d+/\d+|\d+)', q):
                parts = re.split(r'\s*-\s*', q)  # Split on the dash, accounting for possible spaces
                
                # Process part1, check if it's a mixed fraction, pure fraction, or integer
                if re.match(r'\d+\s+\d+/\d+', parts[0]):  # Mixed fraction
                    part1_whole, part1_frac = parts[0].split()
                    part1 = float(part1_whole) + float(Fraction(part1_frac))
                elif '/' in parts[0]:  # Pure fraction
                    part1 = float(Fraction(parts[0]))
                else:  # Integer
                    part1 = float(parts[0])

                # Process part2, same checks as part1
                if re.match(r'\d+\s+\d+/\d+', parts[1]):  # Mixed fraction
                    part2_whole, part2_frac = parts[1].split()
                    part2 = float(part2_whole) + float(Fraction(part2_frac))
                elif '/' in parts[1]:  # Pure fraction
                    part2 = float(Fraction(parts[1]))
                else:  # Integer
                    part2 = float(parts[1])

                # Get the minimum value of the two parts
                min_value = min(part1, part2)
                cleaned_quantities.append(min_value)

            # Handle mixed fractions (e.g., '1 1/2')
            elif re.match(r'\d+\s+\d+/\d+', q):
                parts = q.split()  # Split the whole number and the fraction
                whole_number = float(parts[0])  # Convert the whole number part
                fraction_part = float(Fraction(parts[1]))  # Convert the fraction part
                decimal_value = whole_number + fraction_part  # Sum both parts
                cleaned_quantities.append(decimal_value)

            # Handle pure fractions (e.g., '1/2')
            elif re.match(r'\d+/\d+', q):
                decimal_value = float(Fraction(q))  # Convert the fraction directly to decimal
                cleaned_quantities.append(decimal_value)

            else:
                cleaned_quantities.append(q)  # Leave other string values unchanged
        else:
            cleaned_quantities.append(q)  # Leave non-string values unchanged

    return cleaned_quantities
```

```python
# Apply the cleaning function to the quantity column
projectedIngr['Quantities'] = projectedIngr['RecipeIngredientQuantities'].apply(fix_quantities)
#projectedIngr['Instructions'] = projectedIngr['Instructions'].apply(fix_quantities)

projectedIngr.drop(columns=['RecipeIngredientQuantities'], inplace=True)

projectedIngr.head()
```

### Generate Unit Column

```python
units_list = [
    'teaspoon', 'teaspoons', 'tsp',
    'tablespoon', 'tablespoons', 'tbsp',
    'cup', 'cups',
    'pint', 'pints',
    'quart', 'quarts',
    'gallon', 'gallons',
    'ounce', 'ounces', 'oz',
    'pound', 'pounds', 'lb', 'lbs',
    'gram', 'grams', 'g',
    'kilogram', 'kilograms', 'kg',
    'liter', 'liters', 'l',
    'milliliter', 'milliliters', 'ml',
    'package', 'packages',
    'can', 'cans',
    'slice', 'slices',
    'clove', 'cloves',
    'stick', 'sticks',
    'sprig', 'sprigs',
    'bunch', 'bunches',
    'pinch', 'pinches',
    'dash', 'dashes',
    'head', 'heads',
    'stalk', 'stalks',
    'piece', 'pieces',
    'jar', 'jars',
    'bottle', 'bottles',
    'bag', 'bags',
    'box', 'boxes',
    'block', 'blocks',
    'packet', 'packets',
    'envelope', 'envelopes',
    'sheet', 'sheets',
    'square', 'squares',
    'strip', 'strips',
    'drop', 'drops',
    'fillet', 'fillets',
    'inch', 'inches'
]
```

```python
# Create a regex pattern to match any unit from the list
units_pattern = r'\b(?:' + '|'.join(units_list) + r')\b'
```

```python
import ast  # To safely convert string representation of lists to actual lists

# Function to extract units from the ingredient string
def extract_units(ingredient_list_str):
    ingredient_list = ast.literal_eval(ingredient_list_str)  # Convert string to list
    units = []
    for ingredient in ingredient_list:
        match = re.search(units_pattern, ingredient, re.IGNORECASE)  # Case-insensitive search
        if match:
            units.append(match.group(0).lower())  # Store unit in lowercase for consistency
        else:
            units.append('none')  # Assign 'none' if no unit found
    return units
```

```python
# Apply the function to extract units and create a new 'Unit' column
projectedIngr['Unit'] = projectedIngr['ingredients_raw_str'].apply(extract_units)
```

## Formatting

```python
import numpy as np

# Helper function to convert values to JSON-serializable types
def to_serializable(value):
    if isinstance(value, np.ndarray):
        return value.tolist()  # Convert ndarray to a Python list
    elif isinstance(value, (np.integer, np.floating)):
        return float(value)  # Convert numpy numbers to native floats
    elif isinstance(value, list):
        return [to_serializable(v) for v in value]  # Recursively convert lists
    elif value in [None, 'none', '', 'null']:  # Treat 'none' or missing as None
        return None
    return value  # Return the value as-is for other types
```

```python
# Formatted instructions as a single string rather than a string 
def format_instructions(instructions):
    return ' '.join([f"{i + 1}. {step}" for i, step in enumerate(instructions)])
```

```python
# Updated recipe conversion function
def recipe_to_dict(row):
    ingredients = [
        {
            "name": to_serializable(name),
            "quantity": to_serializable(qty),
            "unit": to_serializable(unit) if unit != 'none' else None
        }
        for name, qty, unit in zip(row['RecipeIngredientParts'], row['Quantities'], row['Unit'])
    ]

    # Build the recipe dictionary
    recipe = {
        "name": to_serializable(row["Name"]),
        "instructions": to_serializable(format_instructions(row["Instructions"])),
        "ingredients": ingredients,
        "total_time": to_serializable(row["TotalTime"]),
        "calories": to_serializable(row["Calories"]),
        "fat": to_serializable(row["FatContent"]),
        "protein": to_serializable(row["ProteinContent"]),
        "carbohydrate": to_serializable(row["CarbohydrateContent"]),
        "desc": row["Description"]
    }

    return recipe
```

```python
import json

recipes = [recipe_to_dict(row) for _, row in projectedIngr.iterrows()]

# Convert to JSON and replace 'null' with 'None'
recipes_json = json.dumps(recipes, indent=4)
recipes_json = recipes_json.replace("null", "\"None\"")

# Save the JSON output to a file
with open("outputRecipes.json", "w") as f:
    f.write(recipes_json)

print("JSON file saved successfully as 'recipes.json'")
```
