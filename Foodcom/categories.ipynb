# Categories of Food.com Dataset

```python
import pandas as pd
```

```python
merged_df = pd.read_parquet('mergedRecipes.parquet')
```

```python
# Count the occurrences of each category
category_counts = merged_df['RecipeCategory'].value_counts()

# Filter the categories that occur more than 100 times
categories_to_keep = category_counts[category_counts > 100].index

# Filter the original dataframe to include only those categories
categoriesProjected = merged_df[merged_df['RecipeCategory'].isin(categories_to_keep)][['Name', 'RecipeIngredientParts', 'RecipeCategory']]
```
