### Read Parquet File

```python
# Use glob to match all parquet files
files = glob.glob('Recipe1M/*.parquet')

# Read and concatenate all files into a single DataFrame
df = pd.concat([pd.read_parquet(file, engine='pyarrow') for file in files], ignore_index=True)
```

```python
df
```

```python
df.columns
```

```python
def count_empty_ingredient(df):
    return df['ingredients'].isna().sum()
```

```python
count_empty_ingredient(df)
```

```python
def remove_duplicates_by_column(df, column):
    # Identify exact duplicate rows based on the 'name' column
    exact_duplicates = df[df.duplicated(subset=column, keep=False)]
    
    # Print the exact duplicates
    print(f"Exact duplicate rows based on {column}:")
    print(exact_duplicates)
    
    # Remove exact duplicates based on the 'name' column, keeping only the first occurrence
    filtered_df = df.drop_duplicates(subset=column, keep='first')
    
    return filtered_df
```

```python
filtered_df_name = remove_duplicates_by_column(df, 'Name')
```

```python
filtered_df_name
```

```python
type(filtered_df_name['Instructions'])
```

```python
filtered_df_name["Concatenated Instructions"] = filtered_df_name["Instructions"].apply(lambda x: " ".join(x))
```

```python
filtered_df_name
```

```python
filtered_df_all = remove_duplicates_by_column(filtered_df_name, 'Concatenated Instructions')
```

```python
filtered_df_all
```

```python
filtered_df_all.to_parquet('deduplicated_recipes_recipe1M.parquet', index=False)

print("Deduplication complete! Results saved to:")
print(" - deduplicated_recipes_recipe1M.parquet (deduplicated data)")
```
