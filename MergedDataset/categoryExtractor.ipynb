```python
import pandas as pd
```

```python
df = pd.read_parquet('outputRecipes.parquet')
```

```python
cat_df = pd.read_parquet('mergedRecipes.parquet')
```

```python
df['RecipeCategory'] = cat_df.RecipeCategory
```

```python
breakfast = ['Breakfast', 'Breakfast Casseroles', 'Breakfast Eggs', 'Brunch', 'Muffins', 'Oatmeal', 'Pancakes', 'Quick Breads']

breads = ['Bread Machine', 'Breads', 'Sourdough Breads', 'Wheat Bread', 'Yeast Breads']

main_dish = ['Bass', 'Bear', 'Beef Liver', 'Beef Organ Meats', 'Catfish', 'Casseroles', 'Chicken', 'Chicken Breast', 'Chicken Crock Pot',
             'Chicken Livers', 'Chicken Thigh & Leg', 'Crab', 'Crawfish', 'Deer', 'Duck', 'Duck Breasts', 'Elk', 'Fish Salmon',
             'Fish Tuna', 'Goose', 'Halibut', 'Lamb/Sheep', 'Lobster', 'Mahi Mahi', 'Main Dish Casseroles', 'Meat', 'Meatballs', 'Meatloaf', 'No Shell Fish',
             'Octopus', 'One Dish Meal', 'Oysters', 'Pheasant', 'Pork', 'Pot Pie', 'Pot Roast', 'Poultry', 'Quail', 'Rabbit', 'Roast', 
             'Roast Beef', 'Roast Beef Crock Pot', 'Savory', 'Squid', 'Steak', 'Tilapia', 'Trout', 'Tuna', 'Turkey Breasts', 'Veal', 
             'Whitefish', 'Whole Chicken', 'Whole Duck', 'Whole Turkey', 'Wild Game']

soup = ['Bean Soup', 'Black Bean Soup', 'Broccoli Soup', 'Chowders', 'Clear Soup', 'Ham And Bean Soup', 'Mushroom Soup',
        'Potato Soup', 'Soups Crock Pot', 'Stew', 'Stocks']

dessert = ['Apple', 'Apple Pie', 'Berries', 'Bread Pudding', 'Candy', 'Cheesecake', 'Cherries', 'Chocolate Chip Cookies', 'Coconut', 
           'Coconut Cream Pie', 'Dessert', 'Desserts Fruit', 'Drop Cookies', 'Frozen Desserts', 'Fruit', 'Grapes', 'Ice Cream', 
           'Jellies', 'Key Lime Pie', 'Kiwifruit', 'Lemon Cake', 'Melons', 'Moose', 'Papaya', 'Peanut Butter', 'Peanut Butter Pie', 
           'Pears', 'Pie', 'Pineapple', 'Plums', 'Raspberries', 'Scones', 'Strawberry', 'Tarts', 'Tropical Fruits']

beverages = ['Beverages', 'Margarita', 'Punch Beverage', 'Shakes', 'Smoothies']

side_dishes = ['Artichoke', 'Avocado', 'Beans', 'Black Beans', 'Brown Rice', 'Cauliflower', 'Chard', 'Cheese', 'Collard Greens', 
               'Corn', 'Greens', 'Grains', 'High Fiber', 'Lentil', 'Long Grain Rice', 'Macaroni And Cheese', 'Mashed Potatoes',
               'Medium Grain Rice', 'Onions', 'Pasta Shells', 'Pennsylvania Dutch', 'Potato', 'Pumpkin', 'Rice', 
               'Salad Dressings', 'Sauces', 'Short Grain Rice', 'Spinach', 'Spreads', 'Sweet Potato/Yam', 'White Rice']

appetizer = ['Appetizers', 'Baking', 'Bar Cookie', 'Chutneys', 'Creole', 'Dip Soslar', 'Gluten Free Appetizers', 'Savory', 
             'Snacks Sweet', 'Spreads']
```

```python
def categorize_recipe_category(category):
    category = str(category)
    if category in breakfast:
        return 'Breakfast'
    elif category in breads:
        return 'Breads'
    elif category in main_dish:
        return 'Main Dish'
    elif category in soup:
        return 'Soup'
    elif category in dessert:
        return 'Dessert'
    elif category in beverages:
        return 'Beverages'
    elif category in side_dishes:
        return 'Side Dishes'
    elif category in appetizer:
        return 'Appetizer'
    else:
        return 'Other'
```

```python
df['MainCategory'] = df['RecipeCategory'].apply(categorize_recipe_category)
```
