```python
import pandas as pd
```

```python
df = pd.read_parquet('outputRecipes.parquet')
```

```python
cat_df = pd.read_parquet('mergedRecipes.parquet')
```

```python
df['RecipeCategory'] = cat_df.RecipeCategory
```

```python
import numpy as np

with open('AllRecipeCategories.txt', 'w') as file:
    for r in np.array(df.RecipeCategory.unique().tolist(), copy=True):
        file.writelines(f'{str(r)}\n')
```

```python
def categorize_recipe_category(category):
    category = str(category)
    if category in ['Breakfast', 'Breakfast Casseroles', 'Breakfast Eggs', 'Brunch', 'Breads', 'Scones', 'Muffins', 'Oatmeal', 'Pancakes', 'Fruit']:
        return 'Breakfast'
    elif category in ['Chicken', 'Poultry', 'Beef Liver', 'Deer', 'Fish Salmon', 'Fish Tuna', 'Duck', 'Pot Pie', 'Roast', 'Steak', 'Veal', 'Rabbit', 'Wild Game', 'Lamb/Sheep', 'Meatballs', 'Meatloaf', 'Pot Roast', 'One Dish Meal', 'Casseroles']:
        return 'Main Dish'
    elif category in ['Bean Soup', 'Black Bean Soup', 'Broccoli Soup', 'Ham And Bean Soup', 'Clear Soup', 'Mushroom Soup', 'Chowders', 'Potato Soup', 'Soups Crock Pot']:
        return 'Soup'
    elif category in ['Dessert', 'Apple Pie', 'Cheesecake', 'Chocolate Chip Cookies', 'Frozen Desserts', 'Pie', 'Coconut Cream Pie', 'Key Lime Pie', 'Peanut Butter Pie', 'Ice Cream', 'Tarts', 'Scones']:
        return 'Dessert'
    elif category in ['Beverages', 'Punch Beverage', 'Shakes', 'Smoothies', 'Margarita', 'Ice Cream']:
        return 'Beverages'
    elif category in ['Broccoli Soup', 'Potato', 'Greens', 'Sauces', 'Salad Dressings', 'Grains', 'Beans', 'Collard Greens', 'Cauliflower', 'Onions', 'Rice']:
        return 'Side Dishes'
    elif category in ['Baking', 'Bar Cookie', 'Snacks Sweet', 'Spreads', 'Dip Soslar', 'Appetizers']:
        return 'Appetizer'
    else:
        return 'Other'
```

```python
df['MainCategory'] = df['RecipeCategory'].apply(categorize_recipe_category)
```
